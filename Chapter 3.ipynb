```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

T = 4.0
N = 10
dt = T / N
tgrid = np.linspace(0, T, N + 1)
numpaths = 500
x = np.zeros((numpaths, N + 1))
x0 = 0.4
x[:, 0] = x0
L = 1.0
d = 10.0e-4

def D(d):
    return np.sqrt(2 * d * dt)

noise_scale = D(d)
for n in range(N):
    xi = np.random.randn(numpaths)
    noise = noise_scale * xi
    
    next_x = x[:, n] + noise
    next_x = np.where(next_x < 0, -next_x, next_x)
    next_x = np.where(next_x > L, (2 * L) - next_x, next_x)
    x[:, n + 1] = next_x

stationary_states = x[:, -1]

sns.set_style('darkgrid')
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6), gridspec_kw={'width_ratios': [2, 1]})

for i in range(numpaths):
    ax1.plot(x[i, :], tgrid, lw=1.2, alpha=0.8)

ax1.axvline(x=L, color='r', linestyle='--', label='Upper Boundary (L)')
ax1.axvline(x=0, color='k', linestyle='--', label='Lower Boundary (0)')
ax1.set_xlim(-0.1, L + 0.1)
ax1.set_xlabel("Position (X)")
ax1.set_ylabel("Time (t)")
ax1.set_title("Confined Brownian Motion Trajectories")
ax1.legend()

sns.histplot(x=stationary_states, kde=True, stat="probability", color="purple", bins=50, alpha=0.6, ax=ax2)
ax2.axvline(x=L, color='r', linestyle='--')
ax2.axvline(x=0, color='k', linestyle='--')
ax2.set_xlabel("Position (X)")
ax2.set_ylabel("Probability Density")
ax2.set_title("Stationary Distribution (t = T)")
ax2.legend()

plt.tight_layout()
plt.show()


```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


def diff(K, N, D, L, t_end):
    h = L / K
    d = D / (h**2)
    
    A = np.zeros(K,dtype=int)
    A[K // 2] = N
    
    t = 0.0
    
    while t < t_end:
        alpha = A * d
        alpha_0 = (2 * d * N) - alpha[0] - alpha[-1]
        
        if alpha_0 <= 0:
            break
            
        r1 = np.random.random()
        t += (1 / alpha_0) * np.log(1 / r1)
        
        r2 = np.random.random()
        right_sum = np.sum(alpha[:-1])
        threshold = right_sum / alpha_0
        
        if r2 < threshold:
            cumulative = np.cumsum(alpha[:-1]) / alpha_0
            j = np.searchsorted(cumulative, r2)
            A[j] -= 1
            A[j+1] += 1
        else:
            cumulative = (right_sum + np.cumsum(alpha[1:])) / alpha_0
            j = np.searchsorted(cumulative, r2) + 1
            A[j] -= 1
            A[j-1] += 1
                    
    return A

final_distribution = diff(K=40, N=1000, D=1e-4, L=1.0, t_end=240)
prob_distribution = final_distribution / np.sum(final_distribution)

# Create the plots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Plot 1: Raw Particle Counts
ax1.bar(range(len(final_distribution)), final_distribution, color='salmon', edgecolor='black')
ax1.set_title("Raw Particle Counts")
ax1.set_xlabel("Lattice Site")
ax1.set_ylabel("Number of Particles")

# Plot 2: Probability Distribution
ax2.bar(range(len(prob_distribution)), prob_distribution, color='teal', edgecolor='black')
ax2.set_title("Probability Distribution")
ax2.set_xlabel("Lattice Site")
ax2.set_ylabel("Probability")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# --- Parameters ---
# We set up the timeframe and simulation resolution
T = 2.0e-11             
N = 1000               
dt = T / N             
D = 1.0e-4                
s = 10**4            
# This rate determines how often our particles switch direction
lambda_rate = (s**2) / (2 * D)  

# Safety check: if our time step is too big, the simulation won't be accurate
if (lambda_rate * dt) >= 0.1:
    print("Heads up: your time step (dt) is quite large; the results might be jittery.")

numpaths = 200         # How many particles we want to track at once

# --- Initialization ---
# Create a matrix to store where each particle is at every point in time
x = np.zeros((numpaths, N + 1))
# Give every particle a random starting direction (either left or right)
v = np.random.choice([-s, s], size=numpaths) 
x[:, 0] = 0.0          # Start everyone at the origin (x=0)

# --- Simulation Loop ---
for n in range(N):
    # Calculate the new position based on current velocity and time step
    x[:, n + 1] = x[:, n] + v * dt
    
    # Now, let's see which particles decide to turn around
    for i in range(numpaths):
        # Toss a digital coin to see if this particle flips
        r = np.random.rand()
        
        # If the random roll is low enough, flip the direction
        # This effectively models the 'velocity-jump' behavior
        if r < (lambda_rate * dt):
            v[i] = -v[i]

# --- Visualization ---
tgrid = np.linspace(0, T, N + 1)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

# Plot the path of the first 20 particles so the graph stays readable
for i in range(min(numpaths, 20)):
    ax1.plot(tgrid, x[i, :], lw=1.0, alpha=0.7)
ax1.set_xlabel("Time (t)")
ax1.set_ylabel("Position (X)")
ax1.set_title("How our particles move over time")
mean_trajectory = np.mean(x, axis=0)
ax1.plot(tgrid, mean_trajectory, color='black', lw=3.0, label='Mean Trajectory')
ax1.legend(loc='upper left')

# Show the final spread of all particles to see the diffusion in action
sns.histplot(x[:, -1], kde=True, stat="probability", color="purple", ax=ax2)
ax2.set_xlabel("Position (X)")
ax2.set_title(f"Where the particles ended up at T={T}")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# --- Parameters ---
T = 1.0                # Total simulation time
D = 0.1                # Diffusion constant
s = 1.0                # Constant speed
lambda_rate = (s**2) / (2 * D)
num_paths = 100        # Number of particles
dt = 0.01              # Uniform sampling interval
time_grid = np.arange(0, T + dt, dt) 
num_steps = len(time_grid)

# Storage for the ensemble: (num_paths, num_steps)
ensemble_x = np.zeros((num_paths, num_steps))

# --- Simulation Loop ---
for i in range(num_paths):
    t = 0.0
    x = 0.0
    v = np.random.choice([-s, s])
    
    # Track the current index in our uniform sampling grid
    step_idx = 0
    ensemble_x[i, step_idx] = x
    
    while t < T:
        # (a11) Generate random number r
        r = np.random.rand()
        
        # (b11) Compute waiting time tau until the next turn
        tau = (1.0 / lambda_rate) * np.log(1.0 / r)
        next_t = t + tau
        
        # (c11) Update positions for all sampling points that fall before the next turn
        while step_idx < num_steps and time_grid[step_idx] < next_t:
            # Particle moves at constant velocity until the turn event
            # We calculate position at each sampling point 'time_grid[step_idx]'
            # relative to the start of this segment 't'
            ensemble_x[i, step_idx] = x + v * (time_grid[step_idx] - t)
            step_idx += 1
            
        # Update simulation time to the turn event
        t = next_t
        x = x + v * tau # Position at the turn
        v = -v          # Direction flip
        
    # Fill remaining steps if any
    ensemble_x[i, step_idx:] = x

# --- Analysis ---
mean_trajectory = np.mean(ensemble_x, axis=0)

# --- Visualization ---
plt.figure(figsize=(10, 6))
# Plot individual paths
for i in range(min(num_paths, 20)):
    plt.plot(time_grid, ensemble_x[i, :], lw=1.0, alpha=0.3, color='gray')

# Plot mean trajectory
plt.plot(time_grid, mean_trajectory, label='Ensemble Mean', color='red', lw=2)

plt.title("Ensemble Mean of Velocity-Jump Trajectories")
plt.xlabel("Time (t)")
plt.ylabel("Position (X)")
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# --- Parameters ---
D = 1e-4          
L = 1.0           
dt = 0.001        
s = 0.1           
lambda_rate = (s**2) / (2 * D)
T = 5.0           
num_steps = int(T / dt)
num_paths = 50    

def simulate_vectorized(n_paths):
    x = np.ones(n_paths) * (0.5 * L)
    v = np.random.choice([-s, s], size=n_paths)
    
    trajectories = np.zeros((n_paths, num_steps + 1))
    trajectories[:, 0] = x
    
    for i in range(1, num_steps + 1):
        x += v * dt
        
        # Correctly capture boundary hits
        hit_low = x < 0
        hit_high = x > L
        
        # Apply reflections: Reassign the results!
        x = np.where(hit_low, -x, x)
        v = np.where(hit_low, -v, v)
        
        x = np.where(hit_high, 2 * L - x, x)
        v = np.where(hit_high, -v, v)
        
        # Stochastic flip: Use rand() for uniform probability 0-1
        flip = np.random.rand(n_paths) < (lambda_rate * dt)
        v = np.where(flip, -v, v)
        
        trajectories[:, i] = x
        
    return trajectories
# Execute and Plot
trajectories=simulate_vectorized(num_paths)
time_grid = np.linspace(0, T, num_steps + 1)

plt.figure(figsize=(10, 6))
for i in range(num_paths):
 plt.plot(time_grid, trajectories[i,:], lw=0.5, alpha=0.3)
mean_trajectory = np.mean(trajectories, axis=0)
# Plot the mean trajectory
plt.plot(time_grid, mean_trajectory, color='black', lw=2.0, label='Mean Trajectory')

# Add the legend to the plot
plt.legend(loc='upper right')
plt.axhline(L, color='red', linestyle='--', label='Boundary')
plt.axhline(0, color='black', linestyle='--')
plt.title("Vectorized Velocity-Jump Simulation")
plt.xlabel("Time (s)")
plt.ylabel("Position (X)")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# --- Parameters ---
D = 10e-4
dt = 0.1
T = 40.0
num_steps = int(T / dt)
num_paths = 100
X_init = 0.6
L = 1.0  # Added Upper Reflecting Boundary

def simulate_vectorized():
    # Rows: paths, Columns: time steps
    trajectories = np.full((num_paths, num_steps + 1), np.nan)
    x = np.full(num_paths, X_init)
    trajectories[:, 0] = x
    
    # Active mask to track trajectories that haven't been absorbed at X=0
    active = np.ones(num_paths, dtype=bool)
    
    for t in range(1, num_steps + 1):
        if not np.any(active):
            break
            
        # (a13) Generate a normally distributed random number xi
        xi = np.random.normal(0, 1, num_paths)
        
        # (b13) Compute the tentative position of the molecule
        x_new = x + np.sqrt(2 * D * dt) * xi
        
        # --- Lower Boundary Logic (Absorbing at X=0) ---
        crossed_zero = x_new < 0
        
        # (d13) Check Brownian bridge probability for hidden hit
        p_hit = np.exp(-(x * x_new) / (D * dt))
        rand_draw = np.random.rand(num_paths)
        hidden_hit = (x_new >= 0) & (x_new <= L) & (rand_draw < p_hit)
        
        # Determine newly absorbed particles
        newly_absorbed = active & (crossed_zero | hidden_hit)
        active = active & ~newly_absorbed
        
        # --- Upper Boundary Logic (Reflecting at X=L) ---
        overshot_L = x_new > L
        reflected_position = 2 * L - x - np.sqrt(2 * D * dt) * xi
        x_new = np.where(overshot_L, reflected_position, x_new)
        
        # --- Finalize Positions ---
        x = np.where(active, x_new, x)
        trajectories[:, t] = np.where(active, x, np.nan)
        
    return trajectories

# --- Run Simulation ---
trajectories = simulate_vectorized()
time_grid = np.linspace(0, T, num_steps + 1)

# --- Plot 1: Trajectories ---
plt.figure(figsize=(10, 6))
plt.plot(time_grid, trajectories.T, lw=0.5, alpha=0.6)
plt.axhline(0, color='red', linestyle='--', label='Absorbing Boundary (X=0)')
plt.axhline(L, color='blue', linestyle='--', label=f'Reflecting Boundary (X={L})')
plt.title("1D Diffusion with Absorbing (X=0) and Reflecting (X=L) Boundaries")
plt.xlabel("Time (s)")
plt.ylabel("Position (X)")
plt.ylim(-0.1, L + 0.1)
plt.legend(loc='upper right')
plt.grid(True, alpha=0.3)
plt.show()

# --- Plot 2: Probability Density at Final Time ---
# Extract the final positions (all paths, last time step), filtering out NaNs
final_data = trajectories[:, -1]
final_data = final_data[~np.isnan(final_data)]

plt.figure(figsize=(8, 5))
sns.histplot(final_data, stat="pro", kde=True, bins=20, color='purple')
plt.title(f"Probability Density of Molecule Positions at T={T}s")
plt.xlabel("Position (X)")
plt.ylabel("Density")
plt.xlim(0, L)
plt.grid(True, alpha=0.3)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# --- Parameters ---
D = 10e-4
dt = 0.1
T = 30.0
num_steps = int(T / dt)
num_paths = 500  # Increased for a clearer distribution
X_init = 0.6
L = 1.0
p_react = 0.3    # NEW: 30% chance to be adsorbed on hit

def simulate_vectorized():
    trajectories = np.full((num_paths, num_steps + 1), np.nan)
    x = np.full(num_paths, X_init)
    trajectories[:, 0] = x
    active = np.ones(num_paths, dtype=bool)
    
    for t in range(1, num_steps + 1):
        if not np.any(active): break
            
        xi = np.random.normal(0, 1, num_paths)
        x_new = x + np.sqrt(2 * D * dt) * xi
        
        # --- Boundary Logic (X=0) ---
        # Detect if the particle hit the boundary
        p_hit = np.exp(-(x * x_new) / (D * dt))
        rand_draw = np.random.rand(num_paths)
        attempted_collision = (x_new < 0) | ((x_new >= 0) & (x_new <= L) & (rand_draw < p_hit))
        
        # Decide: Adsorb or Reflect?
        adsorb_draw = np.random.rand(num_paths)
        is_adsorbed = attempted_collision & (adsorb_draw < p_react)
        
        # Update active mask
        active = active & ~is_adsorbed
        
        # --- Reflection Logic ---
        # If it collided but was NOT adsorbed, it reflects
        # We use np.abs to bounce it back into the domain
        x_new = np.where(attempted_collision & ~is_adsorbed, np.abs(x_new), x_new)
        
        # --- Upper Boundary (X=L) ---
        overshot_L = x_new > L
        x_new = np.where(overshot_L, 2 * L - x_new, x_new)
        
        # Finalize positions
        x = np.where(active, x_new, x)
        trajectories[:, t] = np.where(active, x, np.nan)
        
    return trajectories

# --- Execution and Plotting ---
trajectories = simulate_vectorized()
time_grid = np.linspace(0, T, num_steps + 1)

# Plot 1: Trajectories
plt.figure(figsize=(10, 6))
plt.plot(time_grid, trajectories.T, lw=0.5, alpha=0.5)
plt.axhline(0, color='red', linestyle='--', label=f'Partially Absorbing Boundary (p={p_react})')
plt.axhline(L, color='blue', linestyle='--', label=f'Reflecting Boundary (X={L})')
plt.title(f"Diffusion with Partial Adsorption (p={p_react})")
plt.grid()
plt.legend()
plt.show()

# Plot 2: Final Density
final_data = trajectories[~np.isnan(trajectories[:, -1]), -1]
plt.figure(figsize=(8, 5))
sns.histplot(final_data, stat="probability", bins=20, color='blue',kde='true')
plt.title(f"Spatial Distribution at T={T}s (Partial Adsorption)")
plt.show()
```
